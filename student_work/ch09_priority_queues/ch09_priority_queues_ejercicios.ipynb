StackPriorityQueue

```python
class StackPriorityQueue(UnsortedPriorityQueue):
    def __init__(self):
        self.data = UnsortedPriorityQueue()
        self.key = 0

    def len(self):
        return len(self.data)

    def push(self, e):
        self.data.add(self.key, e)
        self.key -= 1

    def top(self):
        return self.data.min()[1]

    def pop(self):
        return self.data.remove_min()[1]
```

```python
stack = StackPriorityQueue()

stack.push("A")
stack.push("B")
stack.push("C")

print("Cantidad:", stack.len())
print("Cima:", stack.top())
```

```python

print("Pop 1:", stack.pop())
print("Pop 2:", stack.pop())
print("Pop 3:", stack.pop())
```

```python
stack.push("A")
stack.push("B")
stack.push("C")

print("Top:", stack.top())
print("Cantidad después de top:", stack.len())
print("Pop:", stack.pop())
```

EnqueuePriorityQueue

```python
class QueuePriorityQueue:

    def __init__(self):
        self._data = UnsortedPriorityQueue()
        self.key = 0

    def __len__(self):
        return len(self._data)

    def enqueue(self, value):
        self._data.add(self.key, value)
        self.key += 1

    def dequeue(self):
        if self._data.is_empty():
            raise Empty('Queue is empty')
        key, value = self._data.remove_min()

        return value
```

```python
queue = QueuePriorityQueue()

queue.enqueue("A")
queue.enqueue("B")
queue.enqueue("C")

print("Dequeue 1:", queue.dequeue())
print("Dequeue 2:", queue.dequeue())
print("Dequeue 3:", queue.dequeue())
```

```python

queue = QueuePriorityQueue()

print("Antes:", len(queue))

queue.enqueue("A")

print("Después de A:", len(queue))
```

```python
print("Mínimo:", queue._data.min())
```

BinaryTree convertido a lista (Ordenadmiento BFS)

```python

from collections import deque

def por_niveles(tree):
    lista = []

    if tree.is_empty():
        return lista

    cola = deque([tree.root()])

    while cola:
        posicion = cola.popleft()
        lista.append(posicion.element())

        hijo_izquierdo = tree.left(posicion)
        hijo_derecho = tree.right(posicion)

        if hijo_izquierdo is not None:
            cola.append(hijo_izquierdo)

        if hijo_derecho is not None:
            cola.append(hijo_derecho)

    return lista
```

 De lista a binary tree (BFS)

```python


def lista_a_arbol(lista):
    arbol = LinkedBinaryTree()

    if not lista or lista[0] is None:
        return arbol

    posiciones = {0: arbol._add_root(lista[0])}

    for indice, elemento in enumerate(lista):
        if indice == 0 or elemento is None:
            continue

        indice_padre = (indice - 1) // 2
        padre = posiciones.get(indice_padre)

        if padre is None:
            raise ValueError(
                f'El elemento en el indice {indice} no tiene un padre valido'
            )

        if indice % 2 == 1:
            posiciones[indice] = arbol._add_left(padre, elemento)
        else:
            posiciones[indice] = arbol._add_right(padre, elemento)

    return arbol
```

```python
lista = [4, 5, 6, 15, 9]
arbol = lista_a_arbol(lista)

print(por_niveles(arbol))
```

```python
lista_sin_hijos = [4, 5, 6, None, 9]
arbol = lista_a_arbol(lista_sin_hijos)
print(por_niveles(arbol))
```
