```python
from goodrich.ch08.linked_binary_tree import LinkedBinaryTree


def construir_arbol_primos():
    """Construye el árbol de prueba (A-I) usado en los ejercicios de esta sección."""
    T = LinkedBinaryTree()
    a = T._add_root('A')
    b = T._add_left(a, 'B')
    c = T._add_right(a, 'C')
    d = T._add_left(b, 'D')
    e = T._add_right(b, 'E')
    f = T._add_left(c, 'F')
    g = T._add_right(c, 'G')
    h = T._add_left(e, 'H')
    i = T._add_right(g, 'I')
    return T, dict(A=a, B=b, C=c, D=d, E=e, F=f, G=g, H=h, I=i)


# --- Prueba rápida ---
T, N = construir_arbol_primos()
print(len(T))                                       # 9
print(T.root().element())                           # A
print(T.depth(N['H']), T.depth(N['I']))              # 3 3  (mismo nivel)
print(T.sibling(N['D']).element())                   # E
print([c.element() for c in T.children(N['C'])])     # ['F', 'G']
```

# Función que retorna el número de nodos que tienen un hijo (modificable)

```python
def contar_un_hijo(T):
  # Contador que almacena los nodos con hijos
  contador = 0
  # Ciclo que recorre cada uno de los nodos del arbol
  for nodo in T.positions():
    # Comprueba si el nodo tiene determinada cantidad de hijos
    if T.num_children(nodo) == 2:
      # Si tiene determinada cantidad de hijos se suma al contador 1
      contador += 1
  # Retorna la cantidad de nodos con hijos
  return contador
```

```python
contar_un_hijo(T)
```

```python
def contar_un_hijo_derecho(T):
  # Contador que almacena los nodos con hijos
  contador = 0
  # Ciclo que recorre cada uno de los nodos del arbol
  for nodo in T.positions():
    # Comprueba si el nodo No tiene hijo izquierdo
    # si tiene hijo derecho
    if T.left(nodo) is None and T.right(nodo) is not None:
        # Si se cumplen ambas condiciones se suma al contador 1
        contador += 1
  # Retorna la cantidad de nodos que tienen un hijo derecho
  return contador
```

```python
contar_un_hijo_derecho(T)
```

```python
def contar_un_hijo_izquierdo(T):
  contador = 0
  for nodo in T.positions():
    if T.left(nodo) is not None and T.right(nodo) is None:
        contador += 1

  return contador
```

```python
contar_un_hijo_izquierdo(T)
```

# Función que retorna los nodos que tienen un hijo (elementos)

```python
def retornar_nodos_con_un_hijo(T):
  nodos = []
  for nodo in T.positions():
    if T.num_children(nodo) == 2:
      nodos.append(nodo.element())

  return nodos
```

```python
retornar_nodos_con_un_hijo(T)
```

## Función que retorna los hijos de nodos (Izquierda o derecha)

```python
# Función para retornar nodos con hijos a la izquierda
def retornar_hijos_izquierdos(T):
    nodos = []

    for nodo in T.positions():
        if T.left(nodo) is not None and T.right(nodo) is None:
            nodos.append(nodo.element())

    return nodos
```

```python
retornar_hijos_izquierdos(T)
```

```python
# Función para retornar nodos con hijos a la derecha
def retornar_hijos_derechos(T):
    nodos = []

    for nodo in T.positions():
        if T.left(nodo) is None and T.right(nodo) is not None:
            nodos.append(nodo.element())

    return nodos
```

```python
retornar_hijos_derechos(T)
```

# Función que retorna un valor de tipo booleano si los Nodos son  primos

```python
def son_primos(T, nodo1, nodo2):
  # Obtener el nodo del primer padre
    padre1 = T.parent(nodo1)
  # Obtener el nodo del segundo padre
    padre2 = T.parent(nodo2)

    # Si alguno de los nodos no tiene padre, significa que es la raiz del arbol
    # por lo que no puede tener primos
    if padre1 is None or padre2 is None:
        return False

    # Si ambos nodos tienen el mismo padre, significa que son hermnaos y no primos
    if padre1 == padre2:
        return False

    # Comprobamos si los padres de los nodos son hermanos, si es asi ambos nodos
    # son primos
    return T.sibling(padre1) == padre2
```

```python
T, N = construir_arbol_primos()

print(son_primos(T, N['D'], N['F']))
```

# Función que verifica que dos arboles sean iguales

```python
def arboles_iguales(T1, T2):
    # Función auxiliar que compara dos nodos de los arboles
    def comparar(p1, p2):

        # Si los dos nodos son None, significa que ambos árboles
        # tiene un espacio vaio en esa posición por lo tanto son iguales
        if p1 is None and p2 is None:
            return True

        # Si solamente uno de los nodos es None, siginifca que los
        # árboles son diferentes
        if p1 is None or p2 is None:
            return False

        # Comparamos los elementos almacenados en los nodos, si son
        # diferentes, significa que los árboles no son iguales
        if p1.element() != p2.element():
            return False

        # Comparamos el  subárbol izquierdo
        # Si son diferentes retorna False
        if not comparar(T1.left(p1), T2.left(p2)):
            return False

        # Comparamos el subárbol derecho
        # Si son diferentes retorna False
        if not comparar(T1.right(p1), T2.right(p2)):
            return False

        return True
    # Compara los dos arboles desde sus raicez
    return comparar(T1.root(), T2.root())
```

```python
T1, N1 = construir_arbol_primos()
T2, N2 = construir_arbol_primos()

print(arboles_iguales(T1, T2))
```
