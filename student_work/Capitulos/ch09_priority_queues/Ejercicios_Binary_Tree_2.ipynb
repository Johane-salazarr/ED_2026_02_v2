```python
from goodrich.ch08.linked_binary_tree import LinkedBinaryTree


def construir_arbol_primos():
    """Construye el árbol de prueba (A-I) usado en los ejercicios de esta sección."""
    T = LinkedBinaryTree()
    a = T._add_root('A')
    b = T._add_left(a, 'B')
    c = T._add_right(a, 'C')
    d = T._add_left(b, 'D')
    e = T._add_right(b, 'E')
    f = T._add_left(c, 'F')
    g = T._add_right(c, 'G')
    h = T._add_left(e, 'H')
    i = T._add_right(g, 'I')
    return T, dict(A=a, B=b, C=c, D=d, E=e, F=f, G=g, H=h, I=i)


# --- Prueba rápida ---
T, N = construir_arbol_primos()
print(len(T))                                       # 9
print(T.root().element())                           # A
print(T.depth(N['H']), T.depth(N['I']))              # 3 3  (mismo nivel)
print(T.sibling(N['D']).element())                   # E
print([c.element() for c in T.children(N['C'])])     # ['F', 'G']
```

## Ancestros de un nodo

```python
def ancestros(T, nodo):
  # Crear lista vacía donde se almacenan los ancestros del nodo
    resultado = []
  # Obtener el padre del nodo
    padre = T.parent(nodo)
  # Ciclo while hasta que encuentre nodos con padre
    while padre is not None:
  # Agregar a la lista el elemento almacenado en el padre
        resultado.append(padre.element())
  # Subir el nivel en el arbol hasta que encuentre el padre (root)
        padre = T.parent(padre)

    return resultado
```

```python
print(ancestros(T, N["H"]))
```

## Ancestros en común dos nodos

```python
def ancestros_comunes(T, nodo1, nodo2):
    ancestros1 = ancestros(T, nodo1)
    ancestros2 = ancestros(T, nodo2)

    comunes = []

    for nodo in ancestros1:
        if nodo in ancestros2:
            comunes.append(nodo)

    return comunes
```

```python
print(ancestros_comunes(T, N['H'], N['I']))
```

## Función que verifica si un arbol esta completo, si esto es verdadero o falso retorna un valor de tipo booleano

```python
def arbol_completo(T):
  # Si el arbol está vacío, se considera completo
    if len(T) == 0:
        return True
  # Crear una cola que agrega la raiz del arbol
    cola = [T.root()]
  # Variable que indica si se encuentra un nodo con un espacio vacio "sin hijo"
    hueco = False
  # Ciclo que recorre nodos pendientes por revisar
    while cola:
      # Sacar el nodo de la cola
        nodo = cola.pop(0)
      # Obtener los hijos del nodo (convertimos a lista)
        hijos = list(T.children(nodo))
      # Si anteriormente encontramos un hueco y ahora encontramos un nodo con hijos,
      # el árbol no es completo
        if hueco and len(hijos) > 0:
            return False
      # Si el nodo tiene dos hijos, agregamos los hijos a la cola
        if len(hijos) == 2:
            cola.append(hijos[0])
            cola.append(hijos[1])
      # Si el nodo tiene solo un hijo, agregamos el hijo a la cola
        elif len(hijos) == 1:
            cola.append(hijos[0])

            hueco = True
      # Si el nodo no tiene hijos
        else:
      # Encontramos un hueco, por lo que los siguientes nodos
      # no deberian tener hijos
            hueco = True
      # Si al terminar de reccorer el arbol no se encuentra una condición
      # que indique que el árbol no este completo retornamos True
    return True
```

```python
print(arbol_completo(T))
```

## Función que verificar el camino entre dos nodos, retorna la ruta entre los dos nodos

```python
def camino(T, nodo1, nodo2):
    # Creamos una lista que comienza con nodo1
    # Se almacenara el camino desde el nodo1 hasta el nodo2
    camino1 = [nodo1]

    # Almacena temporalmente el nodo1
    actual = nodo1
    # Mientras que el nodo actual tenga un padre,
    # seguimos subiendo de nivel hacia la raiz
    while T.parent(actual) is not None:
      # Padre del nodo actual
        actual = T.parent(actual)
      # Agegar el padre a la lista del camino
        camino1.append(actual)

    # Creamos una lista adicional con nodo2
    # Se almacena el camino desde nodo2 hasta la raiz
    camino2 = [nodo2]
    actual = nodo2

   # Mientras el nodo actual tenga un padre
   # seguimos subiendo de nivel hacia la raiz
    while T.parent(actual) is not None:
      # Padre del nodo actual
        actual = T.parent(actual)
      # Agegar el padre a la lista del camino
        camino2.append(actual)

      # Recorremos todos los nodos del camino desde nodo1 hasta la raíz.
    for i in range(len(camino1)):
      # Recorremos todos los nodos del camino desde nodo2 hasta la raíz.
        for j in range(len(camino2)):
          # Si encontramos un nodo que pertenece a ambos caminos,
            # este será el primer ancestro común.
            if camino1[i] == camino2[j]:

                # Tomamos desde nodo1 hasta el ancestro común,
                # incluyendo el propio ancestro
                ruta = camino1[:i + 1]

                # Tomamos los nodos del camino2 desde el nodo anterior
                # al ancestro común hasta nodo2, pero en orden inverso
                ruta += camino2[j - 1::-1]

                # Convertimos los nodos de la ruta en sus elementos
                return [nodo.element() for nodo in ruta]
    # Si no se encuentra un ancestro común, se retorna la lista vacía
    return []
```

```python
print(camino(T, N['H'], N['I']))
```
